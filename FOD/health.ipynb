## HEALTH HEART EXPERIMENT

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset Health Heart Experimental (health.csv)

```python
# loading the data from dataset serially to the first 5 rows

df = pd.read_csv("health.csv")
df.head(5)
```

## Line Chart

```python
# creating a line chart of HR on x-axis and age on y-axis

df_sorted = df.sort_values('HR')
plt.plot( df_sorted['HR'].head(1000), df_sorted['age'].head(1000), linestyle='-')

# adding a title
plt.title('HR vs age')

# adding a label to the y-axis age and x-axis HR
plt.ylabel('age')
plt.xlabel('HR')
plt.show()
```

## Bar Plot

```python
# bar plotting the heart patients based on their BMI by indication

# adding a title
plt.title("Heart Patients Based on BMI by Heart Indication")

# adding a label to the y-bar BMI and x-bar Indication
df.groupby('indication')['BMI'].mean().plot(kind='bar', color=['blue', 'red'])

plt.xlabel('Indication')
plt.ylabel('Average BMI')
plt.xticks(rotation=0)

plt.show()
```

```python
sns.countplot(data=df, x='sex', hue='indication')
df['indication'].value_counts()
```

```python
sns.barplot(x='SysBP',hue='DiaBP',data=df)
```

## Pie Chart

```python
# pie chart of male and female patients
df['sex'].value_counts().plot(kind='pie',autopct='%0.2f%%')

plt.title('Male vs Female Patients')
```

## Box Plot

```python
# box plotting heart patients
sns.boxplot(x='indication', y='HR', data=df)

# adding a title
plt.title('HR by Indication')
plt.xlabel('Indication (0: No, 1: Yes)')
plt.ylabel('HR')
plt.show()
```

## Histogram

```python
plt.hist(df['weightKg'], edgecolor='black')
```

```python
sns.displot(df['heightCm'], kind="hist", kde=True)
```

```python
sns.kdeplot(df[df['indication'] == 0]['age'], label='No Indication')
sns.kdeplot(df[df['indication'] == 1]['age'], label='Indication')

plt.title('Age Distribution by Indication')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()
plt.show()
```

## Scatter Plot

```python
# Scatter ploting
plt.scatter(df["age"].head(1000), df["HR"].head(1000), marker="o", color="black")


# adding a title
plt.title('Age vs Heart Rate by Heart Indication')

# adding a label to the y-axis HR and x-axis age
plt.xlabel('Age')
plt.ylabel('Heart Rate (bpm)')
plt.show()
```
