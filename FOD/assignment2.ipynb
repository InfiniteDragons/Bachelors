### Matrix Operation

```python
import math
import numpy as np
```

```python
matrixA = np.array([[2, -1], [0, 3]])
matrixB = np.array([[1, 2], [-3, 4]])
print("Matrix A:")
print(matrixA)
print(matrixA.shape)
print("Matrix B:")
print(matrixB)
print(matrixB.shape)

matrixC = matrixA + matrixB
print("Matrix C (A + B):")
print(matrixC)
print(matrixC.shape)

matrixD = matrixA @ matrixB
print("Matrix D (A * B):")
print(matrixD)
print(matrixD.shape)

matrixE = 3*matrixA - 2*matrixB
print("Matrix E (3A - 2B):")
print(matrixE)
print(matrixE.shape)
```

##  Determinants and Inverses  

```python
det = np.linalg.det(matrixA)
print("Determinant of A:")
print(det)

if det != 0:
    matrixF = matrixA - 1
    print("Matrix F (A - 1):")
    print(matrixF)
```

## Eigen Values and Eigen Vectors

```python
matrixC = np.array([[4, 1], [-2, 1]])
print("Matrix C:")
print(matrixC)
print(matrixC.shape)

matrixD = np.linalg.eigh(matrixC)
print("Eigenvalues and Eigenvectors of C:")
print(matrixD)
```

```python
X = np.array([[4, 1.5, 100], [6, 2.0, 150]])
Y = np.array([20, 30])

beta = np.linalg.pinv(X.T @ X) @ X.T @ Y

print("β =", beta)

car3= np.array([[8, 3.0, 200]])

price_car3 = car3 @ beta

print(f" price of car 3 : {price_car3[0]}")
```
