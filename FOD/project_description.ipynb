## **Machine Learning Project – Car Price Prediction**

## Objective 

To build a Linear Regression model that predicts the price of a car based on its features.

## Why Linear Regression was used to make this model?

The target variable (Price) is continuous

We want to understand how different features affect price

Linear Regression is simple, interpretable, and suitable for numeric prediction problems

## Data Collection

Dataset: ALL_CARS_DATA.csv

Total Records: 266,847 cars

## Model Evaluation

from the metrics used the results were:
## **R² Score: 0.718**
Model explains about 71.8% of the variance in car prices.
Strong performance.

## **MSE: 110,893,415**

## **RMSE: 10,530**
this shows that on average, predictions are off by around $10k.



Because Linear Regression performed well:

Car price has a fairly strong linear relationship with:

Model year, Mileage, Age of car, Brand, Car type, Stock type

The dataset structure supports linear modeling reasonably well.

Feature selection and cleaning were effective.

## What This Means Practically

If average car price ≈ $45,000

An average error of $10k is reasonable because:

Luxury cars vary a lot. Some models cost $20k, others $150k+, Price distribution is wide

## **CONCLUSION**

The Linear Regression model achieved an R² score of 0.718, indicating that approximately 71.8% of the variance in car prices is explained by the selected features. The RMSE value of $10,530 suggests a relatively accurate prediction performance considering the wide price range in the dataset. This significant improvement compared to the initial model demonstrates the importance of feature selection, outlier handling, and proper preprocessing.
